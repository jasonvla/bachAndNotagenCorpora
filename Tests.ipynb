```python
zahl1 = float(input("Gib hier die erste Zahl ein "))
zahl2 = float(input("Gib hier die zweite Zahl ein! "))
Operation = str(input("Welche Operation möchtest du ausgeführt haben? Wähle zwischen (+ - * /)"))

if Operation == "+":
    print("Das Ergebnis ist:", zahl1 + zahl2)
elif Operation == "-":
    print("Das Ergebnis ist:", zahl1 - zahl2)
elif Operation == "*":
    print("Das Ergbenis ist:", zahl1 * zahl2)
elif Operation == "/":
    print("Das Ergebnis ist:", zahl1 / zahl2)
else:
    print("Die Eingabe ist leider ungültig!")
```
